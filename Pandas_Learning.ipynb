```python
import pandas as pd

# 1. Create a DataFrame
df = pd.DataFrame({
    "Name": ["Priya", "Vaishu", "Rahul", "Mau"],
    "Joining_Date": [
        "2024-01-15",
        "2023-06-20",
        "2025-03-10",
        "2024-11-05"
    ]
})

print("Original DataFrame:")
print(df)

# 2. Convert text into dates
df["Joining_Date"] = pd.to_datetime(df["Joining_Date"])

print("\nData types:")
print(df.dtypes)

# 3. Extract year, month, and day
df["Year"] = df["Joining_Date"].dt.year
df["Month"] = df["Joining_Date"].dt.month
df["Day"] = df["Joining_Date"].dt.day

# 4. Extract weekday name
df["Day_Name"] = df["Joining_Date"].dt.day_name()

print("\nDataFrame with date details:")
print(df)

# 5. Filter people who joined on or after January 1, 2024
result = df[df["Joining_Date"] >= "2024-01-01"]

print("\nPeople who joined on or after January 1, 2024:")
print(result[["Name", "Joining_Date"]])
```

```python
import pandas as pd

# 1. Create sample data with duplicates
df = pd.DataFrame({
    "Name": ["Priya", "Rahul", "Priya", "Vaishu", "Rahul"],
    "City": ["Pune", "Mumbai", "Pune", "Nashik", "Mumbai"],
    "Marks": [80, 90, 80, 85, 90]
})

print("Original DataFrame:")
print(df)

# 2. Identify duplicate rows
print("\nDuplicate rows:")
print(df.duplicated())

# 3. Count duplicate rows
print("\nNumber of duplicate rows:")
print(df.duplicated().sum())

# 4. Remove duplicate rows
clean_df = df.drop_duplicates()

print("\nDataFrame after removing duplicate rows:")
print(clean_df)

# 5. Check duplicates based only on Name
print("\nDuplicate names:")
print(df.duplicated(subset=["Name"]))

# 6. Remove duplicate names, keeping the last occurrence
df_last = df.drop_duplicates(
    subset=["Name"],
    keep="last"
)

print("\nUnique names, keeping the last occurrence:")
print(df_last)

# 7. Remove every row belonging to a duplicate group
df_no_duplicate_groups = df.drop_duplicates(keep=False)

print("\nRows that were never duplicated:")
print(df_no_duplicate_groups)
```

```python
import pandas as pd

# 1. Create a DataFrame
df = pd.DataFrame({
    "Student_Name": ["Priya", "Vaishu", "Rahul", "Mau"],
    "Student_Age": [20, 22, 21, 23],
    "Student_Marks": [80, 90, 75, 95]
})

print("Original DataFrame:")
print(df)

# 2. Rename one column
df = df.rename(columns={"Student_Name": "Name"})

# 3. Rename multiple columns
df = df.rename(columns={
    "Student_Age": "Age",
    "Student_Marks": "Marks"
})

print("\nAfter renaming columns:")
print(df)

# 4. Rename all columns
df.columns = ["Student", "Age", "Score"]

print("\nAfter changing all column names:")
print(df)

# 5. Rename index labels
df = df.rename(index={
    0: "A",
    1: "B",
    2: "C",
    3: "D"
})

print("\nAfter renaming index labels:")
print(df)

# 6. Reset the index
df = df.reset_index(drop=True)

print("\nAfter resetting the index:")
print(df)

# 7. Set column names using set_axis()
df = df.set_axis(["Name", "Age", "Marks"], axis=1)

print("\nFinal DataFrame:")
print(df)
```

```python
import pandas as pd

# 1. Create a DataFrame
df = pd.DataFrame({
    "Name": ["Priya", "Vaishu", "Rahul", "Mau"],
    "Marks": [80, 90, 45, 65],
    "Age": [20, 22, 21, 23]
})

print("Original DataFrame:")
print(df)

# 2. Convert names to uppercase using map()
df["Name"] = df["Name"].map(str.upper)

print("\nNames in uppercase:")
print(df)

# 3. Add bonus marks using apply()
df["Bonus_Marks"] = df["Marks"].apply(lambda x: x + 5)

# 4. Create a Pass/Fail column
df["Result"] = df["Marks"].apply(
    lambda x: "Pass" if x >= 50 else "Fail"
)

# 5. Create a column for next year's age
df["Age_Next_Year"] = df["Age"].apply(lambda x: x + 1)

# 6. Calculate Marks + Age for each row
df["Total"] = df.apply(
    lambda row: row["Marks"] + row["Age"],
    axis=1
)

print("\nFinal DataFrame:")
print(df)
```

```python
import pandas as pd

# 1. Create the DataFrame
df = pd.DataFrame({
    "Name": ["Priya", "Rahul", "Vaishu", "Mau", "Amit", "Sneha"],
    "City": ["Pune", "Mumbai", "Pune", "Mumbai", "Pune", "Nashik"],
    "Subject": ["Math", "Math", "Science", "Science", "Math", "Science"],
    "Marks": [80, 90, 85, 75, 95, 70]
})

print("Original DataFrame:")
print(df)

# 2. Average marks by city
print("\nAverage marks by city:")
print(pd.pivot_table(
    df,
    values="Marks",
    index="City",
    aggfunc="mean"
))

# 3. Average marks by city and subject
print("\nAverage marks by city and subject:")
print(pd.pivot_table(
    df,
    values="Marks",
    index="City",
    columns="Subject",
    aggfunc="mean"
))

# 4. Replace missing combinations with zero
print("\nAverage marks with missing values filled:")
print(pd.pivot_table(
    df,
    values="Marks",
    index="City",
    columns="Subject",
    aggfunc="mean",
    fill_value=0
))

# 5. Total marks by city
print("\nTotal marks by city:")
print(pd.pivot_table(
    df,
    values="Marks",
    index="City",
    aggfunc="sum"
))

# 6. Count records by city
print("\nRecord count by city:")
print(pd.pivot_table(
    df,
    values="Marks",
    index="City",
    aggfunc="count"
))

# 7. Highest marks by city
print("\nHighest marks by city:")
print(pd.pivot_table(
    df,
    values="Marks",
    index="City",
    aggfunc="max"
))

# 8. Total marks with grand total
print("\nTotal marks including grand total:")
print(pd.pivot_table(
    df,
    values="Marks",
    index="City",
    aggfunc="sum",
    margins=True,
    margins_name="Total"
))
```

```python
import pandas as pd

# 1. Create the DataFrame
df = pd.DataFrame({
    "Name": ["Priya", "Rahul", "Vaishu", "Mau", "Amit", "Sneha", "Riya", "Kunal"],
    "City": ["Pune", "Mumbai", "Pune", "Mumbai", "Pune", "Nashik", "Nashik", "Pune"],
    "Subject": ["Math", "Math", "Science", "Science", "Math", "Science", "Math", "Science"],
    "Gender": ["Female", "Male", "Female", "Female", "Male", "Female", "Female", "Male"]
})

print("Original DataFrame:")
print(df)

# 2. Count students by city and subject
print("\nStudents by City and Subject:")
print(pd.crosstab(df["City"], df["Subject"]))

# 3. Count students with totals
print("\nCity and Subject with Totals:")
print(pd.crosstab(
    df["City"],
    df["Subject"],
    margins=True,
    margins_name="Total"
))

# 4. Percentage within each city
print("\nPercentage by City:")
print(
    (pd.crosstab(
        df["City"],
        df["Subject"],
        normalize="index"
    ) * 100).round(1)
)

# 5. Percentage of all students
print("\nPercentage of All Students:")
print(
    (pd.crosstab(
        df["City"],
        df["Subject"],
        normalize="all"
    ) * 100).round(1)
)

# 6. Count students by city and gender
print("\nStudents by City and Gender:")
print(pd.crosstab(df["City"], df["Gender"]))

# 7. Count students by subject and gender
print("\nStudents by Subject and Gender:")
print(pd.crosstab(df["Subject"], df["Gender"]))
```

```python
import pandas as pd

# 1. Create the dataset
df = pd.DataFrame({
    "Name": ["Priya", "Rahul", "Vaishu", "Amit", "Sneha",
             "Riya", "Kunal", "Mau", "Neha", "Arjun"],
    "City": ["Pune", "Mumbai", "Pune", "Pune", "Nashik",
             "Nashik", "Mumbai", "Mumbai", "Pune", "Nashik"],
    "Subject": ["Math", "Science", "Science", "Math", "Science",
                "Math", "Math", "Science", "Math", "Science"],
    "Marks": [85, 78, 92, 95, 67, 88, 73, None, 81, 90],
    "Attendance": [90, 85, 95, 98, 75, 92, 80, 70, 88, 96]
})

print("1. Original Data:")
print(df)

# 2. Inspect missing values
print("\n2. Missing Values:")
print(df.isnull().sum())

# 3. Fill missing marks with the median
median_marks = df["Marks"].median()
df["Marks"] = df["Marks"].fillna(median_marks)

print("\n3. Median Used:", median_marks)
print(df[["Name", "Marks"]])

# 4. Filter students scoring above 80
print("\n4. Students Scoring Above 80:")
print(df[df["Marks"] > 80][["Name", "Marks"]])

# 5. Average marks by subject
print("\n5. Average Marks by Subject:")
print(df.groupby("Subject")["Marks"].mean())

# 6. Average marks by city
print("\n6. Average Marks by City:")
print(df.groupby("City")["Marks"].mean())

# 7. Average attendance by city
print("\n7. Average Attendance by City:")
print(df.groupby("City")["Attendance"].mean())

# 8. Top five students
print("\n8. Top Five Students:")
print(
    df.sort_values("Marks", ascending=False)
      [["Name", "City", "Marks"]]
      .head(5)
)

# 9. Pivot table: average marks by city and subject
print("\n9. Average Marks by City and Subject:")
print(pd.pivot_table(
    df,
    values="Marks",
    index="City",
    columns="Subject",
    aggfunc="mean"
).round(2))

# 10. Count students by city and subject
print("\n10. Student Counts by City and Subject:")
print(pd.crosstab(df["City"], df["Subject"]))

# 11. Final summary statistics
print("\n11. Overall Statistics:")
print(df[["Marks", "Attendance"]].describe().round(2))
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Create student DataFrame
df = pd.DataFrame({
    "Name": ["Priya", "Rahul", "Vaishu", "Amit", "Sneha"],
    "Marks": [85, 78, 92, 95, 67],
    "Attendance": [90, 85, 95, 98, 75]
})

print("Student Data:")
print(df)

# 2. Vertical bar chart
df.plot(
    x="Name", y="Marks",
    kind="bar", title="Student Marks",
    legend=False
)
plt.xlabel("Student Name")
plt.ylabel("Marks")
plt.xticks(rotation=0)
plt.show()

# 3. Horizontal bar chart
df.plot(
    x="Name", y="Attendance",
    kind="barh", title="Student Attendance",
    legend=False
)
plt.xlabel("Attendance")
plt.ylabel("Student Name")
plt.show()

# 4. Line chart for monthly sales
sales = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May"],
    "Sales": [10000, 15000, 12000, 18000, 22000]
})

sales.plot(
    x="Month", y="Sales",
    kind="line", marker="o",
    title="Monthly Sales",
    legend=False
)
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()

# 5. Histogram of marks
df["Marks"].plot(
    kind="hist", bins=5,
    title="Distribution of Marks"
)
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.show()

# 6. Pie chart for subject proportions
subject_counts = pd.Series({
    "Math": 3,
    "Science": 2
})

subject_counts.plot(
    kind="pie",
    autopct="%1.1f%%",
    title="Students by Subject"
)
plt.ylabel("")
plt.show()

# 7. Scatter plot: attendance versus marks
df.plot(
    x="Attendance", y="Marks",
    kind="scatter",
    title="Marks vs Attendance"
)
plt.xlabel("Attendance")
plt.ylabel("Marks")
plt.show()
```

```python
import pandas as pd

# 1. Create dataset
df = pd.DataFrame({
    "Date": ["2026-01-05", "2026-01-06", "2026-01-07",
             "2026-01-08", "2026-01-09", "2026-01-10"],
    "Product": ["Laptop", "Mouse", "Keyboard",
                "Laptop", "Mouse", "Keyboard"],
    "Category": ["Electronics", "Accessories", "Accessories",
                 "Electronics", "Accessories", "Accessories"],
    "City": ["Pune", "Mumbai", "Pune", "Mumbai", "Pune", "Nashik"],
    "Units": [2, 5, 3, 1, 4, 2],
    "Price": [50000, 500, 1500, 50000, 500, 1500]
})

# 2. Calculate sales
df["Sales"] = df["Units"] * df["Price"]

print("COMPLETE DATASET")
print(df)

# 3. Total sales
print("\nTOTAL SALES")
print(df["Sales"].sum())

# 4. Sales by product
print("\nSALES BY PRODUCT")
product_sales = df.groupby("Product")["Sales"].sum()
print(product_sales)

# 5. Sales by city
print("\nSALES BY CITY")
city_sales = df.groupby("City")["Sales"].sum()
print(city_sales)

# 6. Total units by product
print("\nUNITS SOLD BY PRODUCT")
print(df.groupby("Product")["Units"].sum())

# 7. Detailed product report
print("\nPRODUCT REPORT")
report = df.groupby("Product").agg(
    Total_Sales=("Sales", "sum"),
    Total_Units=("Units", "sum"),
    Average_Price=("Price", "mean")
)
print(report)

# 8. Sort report by total sales
print("\nSORTED SALES REPORT")
print(report.sort_values(by="Total_Sales", ascending=False))

# 9. Pivot table: City vs Product
print("\nCITY-PRODUCT PIVOT TABLE")
pivot = df.pivot_table(
    index="City",
    columns="Product",
    values="Sales",
    aggfunc="sum",
    fill_value=0
)
print(pivot)

# 10. Filter Laptop rows
print("\nLAPTOP SALES DETAILS")
print(df[df["Product"] == "Laptop"])
```

```python
import pandas as pd

# 1. Create sample data
data = {
    "Product": [
        "Laptop", "Mouse", "Keyboard",
        "Laptop", "Mouse", "Keyboard", "Mouse"
    ],
    "City": [
        "Pune", "Mumbai", "Pune",
        "Mumbai", "Pune", "Nashik", "Pune"
    ],
    "Units": [2, 5, 3, 1, None, 2, None],
    "Price": [50000, 500, 1500, 50000, 500, 1500, 500]
}

df = pd.DataFrame(data)

# 2. Save and read CSV
df.to_csv("sales_data.csv", index=False)
df = pd.read_csv("sales_data.csv")

print("ORIGINAL DATA")
print(df.head())

# 3. Check missing values
print("\nMISSING VALUES")
print(df.isnull().sum())

# 4. Remove duplicate rows
print("\nDUPLICATE COUNT:", df.duplicated().sum())
df = df.drop_duplicates()

# 5. Remove rows with missing Units
df = df.dropna(subset=["Units"])

# 6. Clean text columns
df["Product"] = df["Product"].str.strip().str.title()
df["City"] = df["City"].str.strip().str.title()

# 7. Calculate sales
df["Sales"] = df["Units"] * df["Price"]

print("\nCLEANED DATA")
print(df)

# 8. Total sales
print("\nTOTAL SALES")
print(df["Sales"].sum())

# 9. Product-wise report
product_report = df.groupby("Product").agg(
    Total_Units=("Units", "sum"),
    Total_Sales=("Sales", "sum")
)

product_report = product_report.sort_values(
    by="Total_Sales",
    ascending=False
)

print("\nPRODUCT SALES REPORT")
print(product_report)

# 10. City-wise sales
print("\nCITY-WISE SALES")
print(df.groupby("City")["Sales"].sum())

# 11. Save cleaned data and report
df.to_csv("cleaned_sales_data.csv", index=False)
product_report.to_csv("product_sales_report.csv")

print("\nAll CSV files saved successfully!")

# 12. Download files in Google Colab
from google.colab import files

files.download("cleaned_sales_data.csv")
files.download("product_sales_report.csv")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Create dataset
df = pd.DataFrame({
    "Date": [
        "2026-01-05", "2026-01-06", "2026-01-07",
        "2026-01-08", "2026-01-09", "2026-01-10"
    ],
    "Product": [
        "Laptop", "Mouse", "Keyboard",
        "Laptop", "Mouse", "Keyboard"
    ],
    "City": [
        "Pune", "Mumbai", "Pune",
        "Mumbai", "Pune", "Nashik"
    ],
    "Units": [2, 5, 3, 1, 4, 2],
    "Price": [50000, 500, 1500, 50000, 500, 1500]
})

# 2. Prepare data
df["Date"] = pd.to_datetime(df["Date"])
df["Sales"] = df["Units"] * df["Price"]

print("Total Sales:", df["Sales"].sum())

product_sales = df.groupby("Product")["Sales"].sum()
city_sales = df.groupby("City")["Sales"].sum()
daily_sales = df.groupby("Date")["Sales"].sum()

# 3. Product-wise bar chart
product_sales.plot(kind="bar")
plt.title("Product-wise Sales")
plt.xlabel("Product")
plt.ylabel("Sales (Rs.)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig("product_sales_chart.png", dpi=300, bbox_inches="tight")
plt.show()

# 4. City-wise horizontal bar chart
city_sales.plot(kind="barh")
plt.title("City-wise Sales")
plt.xlabel("Sales (Rs.)")
plt.ylabel("City")
plt.tight_layout()
plt.show()

# 5. Daily sales line chart
daily_sales.plot(kind="line", marker="o")
plt.title("Daily Sales Trend")
plt.xlabel("Date")
plt.ylabel("Sales (Rs.)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 6. Units histogram
df["Units"].plot(kind="hist", bins=5, edgecolor="black")
plt.title("Distribution of Units Sold")
plt.xlabel("Units")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

# 7. Product-wise pie chart
product_sales.plot(
    kind="pie",
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Sales Share by Product")
plt.ylabel("")
plt.tight_layout()
plt.show()

# 8. Combined sales dashboard
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

product_sales.plot(kind="bar", ax=axes[0, 0])
axes[0, 0].set_title("Sales by Product")
axes[0, 0].set_xlabel("Product")
axes[0, 0].set_ylabel("Sales (Rs.)")
axes[0, 0].tick_params(axis="x", rotation=0)

city_sales.plot(kind="barh", ax=axes[0, 1])
axes[0, 1].set_title("Sales by City")
axes[0, 1].set_xlabel("Sales (Rs.)")

daily_sales.plot(kind="line", marker="o", ax=axes[1, 0])
axes[1, 0].set_title("Daily Sales Trend")
axes[1, 0].set_xlabel("Date")
axes[1, 0].set_ylabel("Sales (Rs.)")
axes[1, 0].tick_params(axis="x", rotation=45)

df["Units"].plot(
    kind="hist", bins=5, edgecolor="black", ax=axes[1, 1]
)
axes[1, 1].set_title("Units Distribution")
axes[1, 1].set_xlabel("Units")
axes[1, 1].set_ylabel("Frequency")

fig.suptitle("Sales Analysis Dashboard", fontsize=16)
fig.tight_layout()
fig.savefig("sales_dashboard.png", dpi=300, bbox_inches="tight")
plt.show()

print("Dashboard saved as sales_dashboard.png")

# 9. Download dashboard in Google Colab
from google.colab import files
files.download("sales_dashboard.png")
```

```python
import pandas as pd

# 1. Create dataset
df = pd.DataFrame({
    "Employee": ["Amit", "Sneha", "Rahul", "Priya", "Neha", "Rohan"],
    "Department": ["IT", "Sales", "IT", "Sales", "HR", "HR"],
    "City": ["Pune", "Mumbai", "Pune", "Mumbai", "Nashik", "Pune"],
    "Salary": [50000, 40000, 60000, 45000, 35000, 38000],
    "Sales": [80000, 120000, 95000, 70000, 40000, 55000]
})

print("ORIGINAL DATA")
print(df)

# 2. Create Performance column
df["Performance"] = df["Sales"].apply(
    lambda x: "High" if x >= 90000 else "Normal"
)

print("\nPERFORMANCE")
print(df[["Employee", "Sales", "Performance"]])

# 3. Department-wise report
department_report = df.groupby("Department").agg(
    Average_Salary=("Salary", "mean"),
    Total_Sales=("Sales", "sum"),
    Employee_Count=("Employee", "count")
)

print("\nDEPARTMENT REPORT")
print(department_report)

# 4. Employee sales ranking
ranking = df.sort_values(by="Sales", ascending=False)

print("\nSALES RANKING")
print(ranking[["Employee", "Department", "Sales"]])

# 5. Filter using query()
high_sales = df.query("Sales > 70000")

print("\nEMPLOYEES WITH SALES ABOVE 70000")
print(high_sales[["Employee", "Sales"]])

# 6. Filter by multiple conditions
result = df.query("Department == 'IT' and Sales > 90000")

print("\nIT EMPLOYEES WITH SALES ABOVE 90000")
print(result[["Employee", "Sales"]])

# 7. City and department pivot table
pivot = df.pivot_table(
    index="City",
    columns="Department",
    values="Sales",
    aggfunc="sum",
    fill_value=0
)

print("\nCITY-DEPARTMENT PIVOT TABLE")
print(pivot)

# 8. Calculate bonus
df["Bonus"] = df["Sales"].apply(
    lambda x: x * 0.10 if x >= 90000 else x * 0.05
)

print("\nEMPLOYEE BONUS")
print(df[["Employee", "Sales", "Bonus"]])

# 9. Save final report
df.to_csv("employee_analysis_report.csv", index=False)

print("\nReport saved successfully!")
```
